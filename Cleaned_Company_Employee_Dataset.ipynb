```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv")

print("Original Dataset")
print(df.head())

# 2. Inspect the dataset
print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

# 3. Check missing values
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

# 4. Check duplicate records
print("\nDuplicate Records Before Cleaning:")
print(df.duplicated().sum())

# Store original values for comparison
before_rows = df.shape[0]
before_missing = df.isnull().sum().sum()
before_duplicates = df.duplicated().sum()

# 5. Remove completely empty rows
df = df.dropna(how="all")

# 6. Remove duplicate records
df = df.drop_duplicates()

# 7. Clean column names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# 8. Remove unnecessary spaces from text columns
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()

# 9. Standardize inconsistent entries

# Gender
if "gender" in df.columns:
    df["gender"] = df["gender"].replace({
        "M": "Male",
        "m": "Male",
        "male": "Male",
        "F": "Female",
        "f": "Female",
        "female": "Female"
    })

# Department
if "department" in df.columns:
    df["department"] = df["department"].str.upper()

# Work Mode
if "work_mode" in df.columns:
    df["work_mode"] = df["work_mode"].str.title()

# 10. Convert numeric columns to correct data types
numeric_columns = [
    "age",
    "annual_salary",
    "experience_years",
    "performance_score"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# 11. Convert Joining Date to datetime
if "joining_date" in df.columns:
    df["joining_date"] = pd.to_datetime(
        df["joining_date"],
        errors="coerce"
    )

# 12. Fill missing numerical values using median
for col in numeric_columns:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())

# 13. Fill missing categorical values using mode
categorical_columns = [
    "department",
    "gender",
    "city",
    "work_mode"
]

for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mode()[0])

# 14. Forward fill missing dates
if "joining_date" in df.columns:
    df["joining_date"] = df["joining_date"].ffill()

# 15. Check dataset after cleaning
print("\n\nDataset After Cleaning")
print(df.head())

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Records After Cleaning:")
print(df.duplicated().sum())

print("\nFinal Dataset Information:")
df.info()

# 16. Before vs After comparison
after_rows = df.shape[0]
after_missing = df.isnull().sum().sum()
after_duplicates = df.duplicated().sum()

print("\n========== BEFORE vs AFTER ==========")

print("Rows:")
print("Before:", before_rows)
print("After :", after_rows)

print("\nMissing Values:")
print("Before:", before_missing)
print("After :", after_missing)

print("\nDuplicate Records:")
print("Before:", before_duplicates)
print("After :", after_duplicates)

# 17. Export cleaned dataset
df.to_csv("Cleaned_Company_Employee_Dataset.csv", index=False)

print("\nCleaned dataset exported successfully!")
```
