```python
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()
df = pd.read_csv(next(iter(uploaded)))

print("First 5 Rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())

print("\nSelected Columns:")
display(df[['Order_ID', 'Product', 'Category', 'City', 'Total_Sales']].head(10))

print("\nOrders with Sales Greater Than 5000:")
display(df[df['Total_Sales'] > 5000])

print("\nOrders with Quantity Greater Than 3:")
display(df[df['Quantity'] > 3])

print("\nOrders from Chennai:")
display(df[df['City'] == 'Chennai'])

print("\nSports Category Orders:")
display(df[df['Category'] == 'Sports'])

print("\nTop 10 Orders by Sales:")
display(df.sort_values('Total_Sales', ascending=False).head(10))

print("\nTop 10 Orders by Quantity:")
display(df.sort_values('Quantity', ascending=False).head(10))

total_sales = df['Total_Sales'].sum()
average_sales = df['Total_Sales'].mean()
highest_sales = df['Total_Sales'].max()
lowest_sales = df['Total_Sales'].min()
total_quantity = df['Quantity'].sum()
number_of_orders = df['Order_ID'].nunique()

print("\nOverall Sales Statistics")
print("Total Sales:", round(total_sales, 2))
print("Average Sales:", round(average_sales, 2))
print("Highest Sales:", round(highest_sales, 2))
print("Lowest Sales:", round(lowest_sales, 2))
print("Total Quantity Sold:", total_quantity)
print("Number of Orders:", number_of_orders)

category_sales = df.groupby('Category').agg(
    Orders=('Order_ID', 'count'),
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Quantity_Sold=('Quantity', 'sum')
).sort_values('Total_Sales', ascending=False)

print("\nCategory-wise Analysis:")
display(category_sales.round(2))

city_sales = df.groupby('City').agg(
    Orders=('Order_ID', 'count'),
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Quantity_Sold=('Quantity', 'sum')
).sort_values('Total_Sales', ascending=False)

print("\nCity-wise Analysis:")
display(city_sales.round(2))

product_sales = df.groupby('Product').agg(
    Orders=('Order_ID', 'count'),
    Total_Sales=('Total_Sales', 'sum'),
    Quantity_Sold=('Quantity', 'sum')
).sort_values('Total_Sales', ascending=False)

print("\nProduct-wise Analysis:")
display(product_sales.round(2))

print("\nTop 5 Products:")
display(product_sales.head(5).round(2))

payment_sales = df.groupby('Payment_Method').agg(
    Orders=('Order_ID', 'count'),
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean')
).sort_values('Total_Sales', ascending=False)

print("\nPayment Method Analysis:")
display(payment_sales.round(2))

top_category = category_sales.index[0]
top_product = product_sales.index[0]
top_city = city_sales.index[0]
top_payment = payment_sales.index[0]

print("\nMeaningful Observations")
print("1. The dataset contains", number_of_orders, "orders.")
print("2. Total sales generated are ₹", round(total_sales, 2))
print("3. A total of", total_quantity, "items were sold.")
print("4.", top_category, "is the top-performing category by total sales.")
print("5.", top_product, "is the top-performing product by total sales.")
print("6.", top_city, "has the highest total sales among the cities.")
print("7.", top_payment, "generated the highest sales among payment methods.")

plt.figure(figsize=(8, 5))
plt.bar(category_sales.index, category_sales['Total_Sales'])
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
plt.bar(product_sales.head(5).index, product_sales.head(5)['Total_Sales'])
plt.title("Top 5 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
