```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

sns.set_theme(style="whitegrid")

# ============================================================
# 1. LOAD DATASET
# ============================================================

df = pd.read_csv("Day15_Executive_Hotel_Booking_EDA_Dataset.csv")

print("Dataset Shape:", df.shape)

display(df.head())


# ============================================================
# 2. UNDERSTANDING THE DATASET
# ============================================================

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
display(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all").T)


# ============================================================
# 3. MISSING VALUE ANALYSIS
# ============================================================

missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100
})

missing = missing.sort_values(
    "Missing_Percentage",
    ascending=False
)

print("\nMissing Value Analysis:")
display(missing[missing["Missing_Count"] > 0])


# ============================================================
# 4. DUPLICATE RECORDS
# ============================================================

print("\nDuplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape After Removing Duplicates:", df.shape)


# ============================================================
# 5. CATEGORICAL VALUE ANALYSIS
# ============================================================

categorical_columns = df.select_dtypes(
    include="object"
).columns

for col in categorical_columns:
    print("\n", col)
    print(df[col].value_counts(dropna=False).head(20))


# ============================================================
# 6. DATE CONVERSION
# ============================================================

date_columns = [
    "Booking_Date",
    "Arrival_Date",
    "Reservation_Status_Date"
]

for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        errors="coerce"
    )

print("\nDate Columns:")
display(df[date_columns].dtypes)


# ============================================================
# 7. HANDLE MISSING VALUES
# ============================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns

categorical_columns = df.select_dtypes(
    include="object"
).columns

for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(
            df[col].median()
        )

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(
            "Unknown"
        )

print(
    "\nTotal Remaining Missing Values:",
    df.isnull().sum().sum()
)


# ============================================================
# 8. INVALID VALUE CHECK
# ============================================================

print("\nInvalid Value Analysis:")

print(
    "Negative Lead Time:",
    (df["Lead_Time_Days"] < 0).sum()
)

print(
    "Negative Weekend Nights:",
    (df["Weekend_Nights"] < 0).sum()
)

print(
    "Negative Weekday Nights:",
    (df["Weekday_Nights"] < 0).sum()
)

print(
    "Negative Children:",
    (df["Children"] < 0).sum()
)

print(
    "Negative Babies:",
    (df["Babies"] < 0).sum()
)

print(
    "Negative ADR:",
    (df["ADR"] < 0).sum()
)

print(
    "Negative Revenue:",
    (df["Estimated_Revenue"] < 0).sum()
)


# ============================================================
# 9. FEATURE ENGINEERING
# ============================================================

df["Arrival_Year"] = df["Arrival_Date"].dt.year

df["Arrival_Month"] = df["Arrival_Date"].dt.month

df["Arrival_Month_Name"] = (
    df["Arrival_Date"].dt.month_name()
)

df["Arrival_Day"] = (
    df["Arrival_Date"].dt.day
)

df["Arrival_Day_Name"] = (
    df["Arrival_Date"].dt.day_name()
)

df["Total_Guests"] = (
    df["Adults"]
    + df["Children"]
    + df["Babies"]
)

df["Total_Stay_Nights"] = (
    df["Weekend_Nights"]
    + df["Weekday_Nights"]
)

df["Cancellation_Rate"] = (
    df["Is_Canceled"] * 100
)

display(df.head())


# ============================================================
# 10. OUTLIER ANALYSIS
# ============================================================

outlier_columns = [
    "Lead_Time_Days",
    "ADR",
    "Total_Stay_Nights",
    "Total_Guests",
    "Estimated_Revenue"
]

plt.figure(figsize=(15, 10))

for i, col in enumerate(
    outlier_columns, 1
):

    plt.subplot(2, 3, i)

    sns.boxplot(
        y=df[col]
    )

    plt.title(
        f"{col} Outliers"
    )

plt.tight_layout()
plt.show()


# ============================================================
# 11. IQR OUTLIER COUNT
# ============================================================

def outlier_count(data, column):

    Q1 = data[column].quantile(0.25)

    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    return (
        (data[column] < lower)
        |
        (data[column] > upper)
    ).sum()


print("\nOutlier Counts:")

for col in outlier_columns:
    print(
        col,
        ":",
        outlier_count(df, col)
    )


# ============================================================
# 12. HOTEL TYPE DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Hotel_Type"
)

plt.title(
    "Bookings by Hotel Type"
)

plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")

plt.tight_layout()
plt.show()


# ============================================================
# 13. CANCELLATION DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Is_Canceled"
)

plt.title(
    "Booking Cancellation Distribution"
)

plt.xlabel("Cancellation Status")
plt.ylabel("Number of Bookings")

plt.xticks(
    [0, 1],
    ["Not Cancelled", "Cancelled"]
)

plt.tight_layout()
plt.show()


# ============================================================
# 14. LEAD TIME DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Lead_Time_Days",
    bins=40,
    kde=True
)

plt.title(
    "Distribution of Booking Lead Time"
)

plt.xlabel("Lead Time (Days)")
plt.ylabel("Number of Bookings")

plt.tight_layout()
plt.show()


# ============================================================
# 15. ADR DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="ADR",
    bins=40,
    kde=True
)

plt.title(
    "Distribution of Average Daily Rate"
)

plt.xlabel("ADR")
plt.ylabel("Number of Bookings")

plt.tight_layout()
plt.show()


# ============================================================
# 16. TOTAL STAY DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Total_Stay_Nights",
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Total Stay Nights"
)

plt.xlabel("Total Stay Nights")
plt.ylabel("Number of Bookings")

plt.tight_layout()
plt.show()


# ============================================================
# 17. MONTHLY BOOKING TREND
# ============================================================

monthly_bookings = (
    df.groupby(
        df["Arrival_Date"].dt.to_period("M")
    )
    .size()
    .reset_index(
        name="Bookings"
    )
)

monthly_bookings["Month"] = (
    monthly_bookings["Arrival_Date"]
    .astype(str)
)

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_bookings,
    x="Month",
    y="Bookings",
    marker="o"
)

plt.title(
    "Monthly Booking Trend"
)

plt.xlabel("Month")
plt.ylabel("Number of Bookings")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 18. MONTHLY CANCELLATION RATE
# ============================================================

monthly_cancel = (
    df.groupby(
        df["Arrival_Date"].dt.to_period("M")
    )["Is_Canceled"]
    .mean()
    .reset_index()
)

monthly_cancel["Cancellation_Rate"] = (
    monthly_cancel["Is_Canceled"] * 100
)

monthly_cancel["Month"] = (
    monthly_cancel["Arrival_Date"]
    .astype(str)
)

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_cancel,
    x="Month",
    y="Cancellation_Rate",
    marker="o"
)

plt.title(
    "Monthly Cancellation Rate"
)

plt.xlabel("Month")
plt.ylabel(
    "Cancellation Rate (%)"
)

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 19. MARKET SEGMENT ANALYSIS
# ============================================================

segment_analysis = (
    df.groupby("Market_Segment")
    .agg(
        Bookings=(
            "Booking_ID",
            "count"
        ),
        Cancellation_Rate=(
            "Is_Canceled",
            "mean"
        ),
        Average_ADR=(
            "ADR",
            "mean"
        ),
        Revenue=(
            "Estimated_Revenue",
            "sum"
        )
    )
    .sort_values(
        "Bookings",
        ascending=False
    )
)

segment_analysis[
    "Cancellation_Rate"
] *= 100

display(segment_analysis)


# ============================================================
# 20. MARKET SEGMENT VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Market_Segment",
    order=df[
        "Market_Segment"
    ].value_counts().index
)

plt.title(
    "Bookings by Market Segment"
)

plt.xlabel(
    "Number of Bookings"
)

plt.ylabel(
    "Market Segment"
)

plt.tight_layout()
plt.show()


# ============================================================
# 21. HOTEL TYPE VS CANCELLATION
# ============================================================

hotel_cancel = (
    df.groupby("Hotel_Type")
    ["Is_Canceled"]
    .mean()
    .mul(100)
    .reset_index()
)

hotel_cancel.columns = [
    "Hotel_Type",
    "Cancellation_Rate"
]

display(hotel_cancel)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=hotel_cancel,
    x="Hotel_Type",
    y="Cancellation_Rate"
)

plt.title(
    "Cancellation Rate by Hotel Type"
)

plt.xlabel("Hotel Type")
plt.ylabel(
    "Cancellation Rate (%)"
)

plt.tight_layout()
plt.show()


# ============================================================
# 22. HOTEL TYPE VS REVENUE
# ============================================================

hotel_revenue = (
    df.groupby("Hotel_Type")
    .agg(
        Total_Revenue=(
            "Estimated_Revenue",
            "sum"
        ),
        Average_Revenue=(
            "Estimated_Revenue",
            "mean"
        ),
        Average_ADR=(
            "ADR",
            "mean"
        )
    )
)

display(hotel_revenue)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Hotel_Type",
    y="Estimated_Revenue",
    estimator="sum"
)

plt.title(
    "Estimated Revenue by Hotel Type"
)

plt.xlabel("Hotel Type")
plt.ylabel(
    "Estimated Revenue"
)

plt.tight_layout()
plt.show()


# ============================================================
# 23. LEAD TIME VS CANCELLATION
# ============================================================

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Is_Canceled",
    y="Lead_Time_Days"
)

plt.title(
    "Lead Time vs Cancellation"
)

plt.xlabel(
    "Cancellation Status"
)

plt.ylabel(
    "Lead Time (Days)"
)

plt.xticks(
    [0, 1],
    ["Not Cancelled", "Cancelled"]
)

plt.tight_layout()
plt.show()


# ============================================================
# 24. LEAD TIME CATEGORY
# ============================================================

df["Lead_Time_Category"] = pd.cut(
    df["Lead_Time_Days"],
    bins=[
        -1,
        7,
        30,
        90,
        180,
        np.inf
    ],
    labels=[
        "0-7 Days",
        "8-30 Days",
        "31-90 Days",
        "91-180 Days",
        "180+ Days"
    ]
)

lead_cancel = (
    df.groupby(
        "Lead_Time_Category",
        observed=True
    )
    .agg(
        Bookings=(
            "Booking_ID",
            "count"
        ),
        Cancellation_Rate=(
            "Is_Canceled",
            "mean"
        )
    )
)

lead_cancel[
    "Cancellation_Rate"
] *= 100

display(lead_cancel)


plt.figure(figsize=(10, 6))

sns.barplot(
    data=lead_cancel.reset_index(),
    x="Lead_Time_Category",
    y="Cancellation_Rate"
)

plt.title(
    "Cancellation Rate by Lead-Time Category"
)

plt.xlabel(
    "Lead-Time Category"
)

plt.ylabel(
    "Cancellation Rate (%)"
)

plt.tight_layout()
plt.show()


# ============================================================
# 25. CUSTOMER TYPE ANALYSIS
# ============================================================

customer_analysis = (
    df.groupby("Customer_Type")
    .agg(
        Bookings=(
            "Booking_ID",
            "count"
        ),
        Cancellation_Rate=(
            "Is_Canceled",
            "mean"
        ),
        Average_ADR=(
            "ADR",
            "mean"
        ),
        Average_Stay=(
            "Total_Stay_Nights",
            "mean"
        ),
        Revenue=(
            "Estimated_Revenue",
            "sum"
        )
    )
)

customer_analysis[
    "Cancellation_Rate"
] *= 100

display(
    customer_analysis.sort_values(
        "Bookings",
        ascending=False
    )
)


# ============================================================
# 26. CUSTOMER TYPE VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Customer_Type",
    order=df[
        "Customer_Type"
    ].value_counts().index
)

plt.title(
    "Bookings by Customer Type"
)

plt.xlabel(
    "Number of Bookings"
)

plt.ylabel(
    "Customer Type"
)

plt.tight_layout()
plt.show()


# ============================================================
# 27. ROOM TYPE ANALYSIS
# ============================================================

room_analysis = (
    df.groupby("Room_Type_Reserved")
    .agg(
        Bookings=(
            "Booking_ID",
            "count"
        ),
        Cancellation_Rate=(
            "Is_Canceled",
            "mean"
        ),
        Average_ADR=(
            "ADR",
            "mean"
        ),
        Revenue=(
            "Estimated_Revenue",
            "sum"
        )
    )
    .sort_values(
        "Bookings",
        ascending=False
    )
)

room_analysis[
    "Cancellation_Rate"
] *= 100

display(room_analysis)


# ============================================================
# 28. ROOM TYPE VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Room_Type_Reserved",
    order=df[
        "Room_Type_Reserved"
    ].value_counts().index
)

plt.title(
    "Bookings by Reserved Room Type"
)

plt.xlabel(
    "Number of Bookings"
)

plt.ylabel(
    "Room Type"
)

plt.tight_layout()
plt.show()


# ============================================================
# 29. MEAL TYPE ANALYSIS
# ============================================================

meal_analysis = (
    df.groupby("Meal_Type")
    .agg(
        Bookings=(
            "Booking_ID",
            "count"
        ),
        Average_ADR=(
            "ADR",
            "mean"
        ),
        Cancellation_Rate=(
            "Is_Canceled",
            "mean"
        )
    )
)

meal_analysis[
    "Cancellation_Rate"
] *= 100

display(meal_analysis)


# ============================================================
# 30. REPEAT GUEST ANALYSIS
# ============================================================

repeat_analysis = (
    df.groupby("Is_Repeated_Guest")
    .agg(
        Bookings=(
            "Booking_ID",
            "count"
        ),
        Cancellation_Rate=(
            "Is_Canceled",
            "mean"
        ),
        Average_ADR=(
            "ADR",
            "mean"
        ),
        Average_Stay=(
            "Total_Stay_Nights",
            "mean"
        ),
        Revenue=(
            "Estimated_Revenue",
            "sum"
        )
    )
)

repeat_analysis[
    "Cancellation_Rate"
] *= 100

display(repeat_analysis)


# ============================================================
# 31. DISTRIBUTION CHANNEL ANALYSIS
# ============================================================

channel_analysis = (
    df.groupby("Distribution_Channel")
    .agg(
        Bookings=(
            "Booking_ID",
            "count"
        ),
        Cancellation_Rate=(
            "Is_Canceled",
            "mean"
        ),
        Average_ADR=(
            "ADR",
            "mean"
        ),
        Revenue=(
            "Estimated_Revenue",
            "sum"
        )
    )
    .sort_values(
        "Bookings",
        ascending=False
    )
)

channel_analysis[
    "Cancellation_Rate"
] *= 100

display(channel_analysis)


# ============================================================
# 32. COUNTRY ANALYSIS
# ============================================================

country_analysis = (
    df.groupby("Country")
    .agg(
        Bookings=(
            "Booking_ID",
            "count"
        ),
        Cancellation_Rate=(
            "Is_Canceled",
            "mean"
        ),
        Revenue=(
            "Estimated_Revenue",
            "sum"
        )
    )
    .sort_values(
        "Bookings",
        ascending=False
    )
)

country_analysis[
    "Cancellation_Rate"
] *= 100

display(
    country_analysis.head(15)
)


# ============================================================
# 33. TOP COUNTRIES
# ============================================================

top_countries = (
    df["Country"]
    .value_counts()
    .head(15)
    .index
)

plt.figure(figsize=(12, 7))

sns.countplot(
    data=df[
        df["Country"].isin(
            top_countries
        )
    ],
    y="Country",
    order=top_countries
)

plt.title(
    "Top 15 Countries by Number of Bookings"
)

plt.xlabel(
    "Number of Bookings"
)

plt.ylabel("Country")

plt.tight_layout()
plt.show()


# ============================================================
# 34. SPECIAL REQUESTS
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Total_Special_Requests"
)

plt.title(
    "Distribution of Special Requests"
)

plt.xlabel(
    "Number of Special Requests"
)

plt.ylabel(
    "Number of Bookings"
)

plt.tight_layout()
plt.show()


# ============================================================
# 35. SATISFACTION SCORE
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Satisfaction_Score",
    bins=20,
    kde=True
)

plt.title(
    "Distribution of Guest Satisfaction Score"
)

plt.xlabel(
    "Satisfaction Score"
)

plt.ylabel(
    "Number of Bookings"
)

plt.tight_layout()
plt.show()


# ============================================================
# 36. SATISFACTION VS ADR
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Satisfaction_Score",
    y="ADR",
    hue="Hotel_Type",
    alpha=0.6
)

plt.title(
    "Satisfaction Score vs ADR"
)

plt.xlabel(
    "Satisfaction Score"
)

plt.ylabel("ADR")

plt.tight_layout()
plt.show()


# ============================================================
# 37. CORRELATION HEATMAP
# ============================================================

numeric_df = df.select_dtypes(
    include=np.number
)

correlation = numeric_df.corr()

plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Heatmap"
)

plt.tight_layout()
plt.show()


# ============================================================
# 38. CORRELATION WITH CANCELLATION
# ============================================================

cancellation_correlation = (
    correlation["Is_Canceled"]
    .sort_values(
        ascending=False
    )
)

print(
    "\nCorrelation with Cancellation:"
)

display(
    cancellation_correlation
)


# ============================================================
# 39. REVENUE BY MARKET SEGMENT
# ============================================================

segment_revenue = (
    df.groupby(
        "Market_Segment"
    )["Estimated_Revenue"]
    .sum()
    .sort_values(
        ascending=False
    )
)

plt.figure(figsize=(10, 6))

segment_revenue.plot(
    kind="bar"
)

plt.title(
    "Revenue by Market Segment"
)

plt.xlabel(
    "Market Segment"
)

plt.ylabel(
    "Estimated Revenue"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()
plt.show()


# ============================================================
# 40. EXECUTIVE KPIs
# ============================================================

total_bookings = len(df)

total_revenue = (
    df["Estimated_Revenue"].sum()
)

average_adr = (
    df["ADR"].mean()
)

cancellation_rate = (
    df["Is_Canceled"].mean() * 100
)

average_stay = (
    df["Total_Stay_Nights"].mean()
)

repeat_guest_rate = (
    df["Is_Repeated_Guest"].mean() * 100
)

print("\n")
print("=" * 55)
print("          EXECUTIVE HOTEL PERFORMANCE")
print("=" * 55)

print(
    f"Total Bookings       : {total_bookings:,}"
)

print(
    f"Estimated Revenue    : {total_revenue:,.2f}"
)

print(
    f"Average ADR          : {average_adr:,.2f}"
)

print(
    f"Cancellation Rate    : {cancellation_rate:.2f}%"
)

print(
    f"Average Stay         : {average_stay:.2f} nights"
)

print(
    f"Repeat Guest Rate    : {repeat_guest_rate:.2f}%"
)

print("=" * 55)


# ============================================================
# 41. EXECUTIVE DASHBOARD
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(15, 10)
)

sns.countplot(
    data=df,
    x="Hotel_Type",
    ax=axes[0, 0]
)

axes[0, 0].set_title(
    "Bookings by Hotel Type"
)

sns.barplot(
    data=hotel_cancel,
    x="Hotel_Type",
    y="Cancellation_Rate",
    ax=axes[0, 1]
)

axes[0, 1].set_title(
    "Cancellation Rate by Hotel Type"
)

axes[0, 1].set_ylabel(
    "Cancellation Rate (%)"
)

sns.histplot(
    data=df,
    x="ADR",
    bins=40,
    kde=True,
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    "ADR Distribution"
)

sns.countplot(
    data=df,
    y="Market_Segment",
    order=df[
        "Market_Segment"
    ].value_counts().index,
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    "Bookings by Market Segment"
)

plt.tight_layout()

plt.show()


# ============================================================
# 42. AUTOMATIC BUSINESS INSIGHTS
# ============================================================

highest_revenue_segment = (
    df.groupby(
        "Market_Segment"
    )["Estimated_Revenue"]
    .sum()
    .idxmax()
)

highest_booking_segment = (
    df["Market_Segment"]
    .value_counts()
    .idxmax()
)

highest_cancel_hotel = (
    df.groupby(
        "Hotel_Type"
    )["Is_Canceled"]
    .mean()
    .idxmax()
)

highest_cancel_lead = (
    df.groupby(
        "Lead_Time_Category",
        observed=True
    )["Is_Canceled"]
    .mean()
    .idxmax()
)

highest_adr_hotel = (
    df.groupby(
        "Hotel_Type"
    )["ADR"]
    .mean()
    .idxmax()
)

print("\n")
print("=" * 60)
print("                 KEY BUSINESS INSIGHTS")
print("=" * 60)

print(
    "1. Highest booking-volume market segment:",
    highest_booking_segment
)

print(
    "2. Highest revenue-generating market segment:",
    highest_revenue_segment
)

print(
    "3. Hotel type with highest cancellation rate:",
    highest_cancel_hotel
)

print(
    "4. Hotel type with highest average ADR:",
    highest_adr_hotel
)

print(
    "5. Lead-time group with highest cancellation:",
    highest_cancel_lead
)

print("=" * 60)


# ============================================================
# 43. MANAGEMENT RECOMMENDATIONS
# ============================================================

recommendations = pd.DataFrame({

    "Recommendation": [

        "Introduce stronger confirmation policies for high-risk bookings.",

        "Use lead-time based cancellation risk rules.",

        "Focus marketing investment on high-revenue market segments.",

        "Implement dynamic pricing based on demand, hotel type and booking behaviour.",

        "Develop loyalty benefits for repeat guests.",

        "Use guest preferences and special requests to personalize services.",

        "Monitor cancellation, ADR and revenue KPIs through a recurring management dashboard."
    ],

    "Expected_Benefit": [

        "Reduce avoidable cancellations and revenue loss.",

        "Improve booking predictability and occupancy planning.",

        "Increase marketing effectiveness.",

        "Improve room revenue and pricing efficiency.",

        "Increase customer retention.",

        "Improve guest satisfaction.",

        "Enable faster management decisions."
    ]
})

display(
    recommendations
)


# ============================================================
# 44. FINAL DATA QUALITY CHECK
# ============================================================

print("\n")
print("=" * 60)
print("              FINAL DATA QUALITY CHECK")
print("=" * 60)

print(
    "Final Dataset Shape:",
    df.shape
)

print(
    "Duplicate Rows:",
    df.duplicated().sum()
)

print(
    "Total Missing Values:",
    df.isnull().sum().sum()
)

display(
    df.isnull().sum()
    .sort_values(
        ascending=False
    )
    .head(20)
)


# ============================================================
# 45. EXPORT CLEANED DATASET
# ============================================================

df.to_csv(
    "Cleaned_Hotel_Booking_EDA_Dataset.csv",
    index=False
)

print(
    "\nCleaned dataset saved successfully."
)

print(
    "Cleaned_Hotel_Booking_EDA_Dataset.csv"
)
```
