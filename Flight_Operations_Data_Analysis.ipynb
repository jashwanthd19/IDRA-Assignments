```python
import pandas as pd

df = pd.read_csv("Day10_Flight_Operations_Dataset.csv")

print("FIRST 5 ROWS")
print(df.head())

print("\nLAST 5 ROWS")
print(df.tail())

print("\nRANDOM SAMPLE")
print(df.sample(5))

print("\nSHAPE")
print(df.shape)

print("\nCOLUMNS")
print(df.columns)

print("\nDATASET INFORMATION")
df.info()

print("\nSTATISTICAL SUMMARY")
print(df.describe())

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])

df["Delay_Minutes"] = df["Delay_Minutes"].fillna(0)

print("\nSELECTED COLUMNS")
print(df[[
    "Flight_ID",
    "Airline",
    "Origin",
    "Destination",
    "Passengers"
]].head())

print("\nFLIGHTS WITH MORE THAN 150 PASSENGERS")
print(df[df["Passengers"] > 150][[
    "Flight_ID",
    "Airline",
    "Passengers",
    "Flight_Status"
]])

print("\nFLIGHTS DELAYED MORE THAN 30 MINUTES")
print(df[df["Delay_Minutes"] > 30][[
    "Flight_ID",
    "Airline",
    "Delay_Minutes",
    "Weather"
]])

print("\nTOP 10 FLIGHTS BY DELAY")
print(df.sort_values(
    "Delay_Minutes",
    ascending=False
)[[
    "Flight_ID",
    "Airline",
    "Delay_Minutes",
    "Flight_Status"
]].head(10))

print("\nTOP 10 FLIGHTS BY PASSENGER SATISFACTION")
print(df.sort_values(
    "Passenger_Satisfaction",
    ascending=False
)[[
    "Flight_ID",
    "Airline",
    "Passenger_Satisfaction"
]].head(10))

airline_analysis = df.groupby("Airline").agg(
    Flights=("Flight_ID", "count"),
    Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values("Flights", ascending=False)

print("\nAIRLINE ANALYSIS")
print(airline_analysis)

print("\nFLIGHT STATUS")
print(df["Flight_Status"].value_counts())

weather_analysis = df.groupby("Weather").agg(
    Flights=("Flight_ID", "count"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values("Average_Delay", ascending=False)

print("\nWEATHER ANALYSIS")
print(weather_analysis)

class_analysis = df.groupby("Travel_Class").agg(
    Flights=("Flight_ID", "count"),
    Passengers=("Passengers", "sum"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
)

print("\nTRAVEL CLASS ANALYSIS")
print(class_analysis)

booking_analysis = df.groupby("Booking_Channel").agg(
    Flights=("Flight_ID", "count"),
    Passengers=("Passengers", "sum"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values("Flights", ascending=False)

print("\nBOOKING CHANNEL ANALYSIS")
print(booking_analysis)

route_analysis = df.groupby([
    "Origin",
    "Destination"
]).agg(
    Flights=("Flight_ID", "count"),
    Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean")
).sort_values("Flights", ascending=False)

print("\nTOP 10 ROUTES")
print(route_analysis.head(10))

df["Month"] = df["Flight_Date"].dt.month_name()
df["Day"] = df["Flight_Date"].dt.day
df["Day_Name"] = df["Flight_Date"].dt.day_name()

monthly_analysis = df.groupby("Month").agg(
    Flights=("Flight_ID", "count"),
    Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
)

print("\nMONTHLY ANALYSIS")
print(monthly_analysis)

df["Occupancy_Rate"] = (
    df["Passengers"] / df["Seat_Capacity"] * 100
).round(2)

df["Delay_Category"] = df["Delay_Minutes"].apply(
    lambda x: "On Time"
    if x == 0
    else ("Minor Delay" if x <= 30 else "Major Delay")
)

df["Estimated_Revenue"] = (
    df["Passengers"] * df["Average_Ticket_Price"]
)

print("\nTRANSFORMED DATA")
print(df[[
    "Flight_ID",
    "Flight_Date",
    "Airline",
    "Passengers",
    "Seat_Capacity",
    "Occupancy_Rate",
    "Delay_Minutes",
    "Delay_Category",
    "Estimated_Revenue"
]].head(10))

print("\nDELAY CATEGORY ANALYSIS")
print(df["Delay_Category"].value_counts())

print("\nHIGHEST PASSENGER FLIGHT")
print(df.loc[
    df["Passengers"].idxmax(),
    [
        "Flight_ID",
        "Airline",
        "Passengers",
        "Seat_Capacity"
    ]
])

print("\nHIGHEST DELAY FLIGHT")
print(df.loc[
    df["Delay_Minutes"].idxmax(),
    [
        "Flight_ID",
        "Airline",
        "Delay_Minutes",
        "Weather"
    ]
])

print("\nHIGHEST TICKET PRICE")
print(df.loc[
    df["Average_Ticket_Price"].idxmax(),
    [
        "Flight_ID",
        "Airline",
        "Average_Ticket_Price"
    ]
])

print("\nAVERAGE OCCUPANCY RATE")
print(round(df["Occupancy_Rate"].mean(), 2))

print("\nOVERALL AVERAGE DELAY")
print(round(df["Delay_Minutes"].mean(), 2))

print("\nOVERALL PASSENGER SATISFACTION")
print(round(df["Passenger_Satisfaction"].mean(), 2))

print("\nFINAL DATASET SHAPE")
print(df.shape)
```
