```python
# ============================================================
# ML STUDENT PERFORMANCE PREDICTION AND MODEL EVALUATION
# Day 18 & Day 19
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import warnings
warnings.filterwarnings("ignore")


# ============================================================
# 2. LOAD DATASET
# ============================================================

import pandas as pd

df = pd.read_csv("Day18_19_student_habits_performance.csv")

print("Dataset loaded successfully!")

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 3. BASIC DATASET INFORMATION
# ============================================================

print("\nShape of dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()


# ============================================================
# 4. STATISTICAL SUMMARY
# ============================================================

print("\nStatistical Summary:")
print(df.describe())


print("\nCategorical Variable Summary:")
print(df.describe(include="object"))


# ============================================================
# 5. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": missing_percentage
})

print("\nMissing Value Summary:")
print(missing_table)


# ============================================================
# 6. IDENTIFY NUMERICAL AND CATEGORICAL VARIABLES
# ============================================================

numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_columns = df.select_dtypes(
    include=["object"]
).columns

print("\nNumerical Variables:")
print(list(numerical_columns))

print("\nCategorical Variables:")
print(list(categorical_columns))


# ============================================================
# 7. UNIQUE VALUES OF CATEGORICAL VARIABLES
# ============================================================

print("\nUnique values in categorical variables:")

for column in categorical_columns:

    print("\n--------------------------------")
    print(column)

    print(df[column].value_counts(dropna=False))


# ============================================================
# 8. DISTRIBUTION OF EXAM SCORE
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    df["exam_score"],
    kde=True
)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.show()


# ============================================================
# 9. BOXPLOT OF EXAM SCORE
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["exam_score"]
)

plt.title("Boxplot of Exam Scores")
plt.xlabel("Exam Score")

plt.show()


# ============================================================
# 10. DISTRIBUTION OF NUMERICAL VARIABLES
# ============================================================

df[numerical_columns].hist(
    figsize=(15, 12),
    bins=20
)

plt.tight_layout()
plt.show()


# ============================================================
# 11. CORRELATION MATRIX
# ============================================================

correlation_matrix = df[numerical_columns].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()


# ============================================================
# 12. CORRELATION WITH EXAM SCORE
# ============================================================

exam_correlation = (
    correlation_matrix["exam_score"]
    .sort_values(ascending=False)
)

print("\nCorrelation with Exam Score:")
print(exam_correlation)


# ============================================================
# 13. SORT VARIABLES BY ABSOLUTE CORRELATION
# ============================================================

exam_correlation_absolute = (
    correlation_matrix["exam_score"]
    .drop("exam_score")
    .sort_values(
        key=abs,
        ascending=False
    )
)

print("\nVariables most strongly related to exam_score:")
print(exam_correlation_absolute)


# ============================================================
# 14. STUDY HOURS VS EXAM SCORE
# ============================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="study_hours_per_day",
    y="exam_score"
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours per Day")
plt.ylabel("Exam Score")

plt.show()


# ============================================================
# 15. ATTENDANCE VS EXAM SCORE
# ============================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="attendance_percentage",
    y="exam_score"
)

plt.title("Attendance Percentage vs Exam Score")
plt.xlabel("Attendance Percentage")
plt.ylabel("Exam Score")

plt.show()


# ============================================================
# 16. SLEEP HOURS VS EXAM SCORE
# ============================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="sleep_hours",
    y="exam_score"
)

plt.title("Sleep Hours vs Exam Score")
plt.xlabel("Sleep Hours")
plt.ylabel("Exam Score")

plt.show()


# ============================================================
# 17. SOCIAL MEDIA HOURS VS EXAM SCORE
# ============================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="social_media_hours",
    y="exam_score"
)

plt.title("Social Media Hours vs Exam Score")
plt.xlabel("Social Media Hours")
plt.ylabel("Exam Score")

plt.show()


# ============================================================
# 18. NETFLIX HOURS VS EXAM SCORE
# ============================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="netflix_hours",
    y="exam_score"
)

plt.title("Netflix Hours vs Exam Score")
plt.xlabel("Netflix Hours")
plt.ylabel("Exam Score")

plt.show()


# ============================================================
# 19. MENTAL HEALTH RATING VS EXAM SCORE
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="mental_health_rating",
    y="exam_score"
)

plt.title("Mental Health Rating vs Exam Score")
plt.xlabel("Mental Health Rating")
plt.ylabel("Exam Score")

plt.show()


# ============================================================
# 20. EXERCISE FREQUENCY VS EXAM SCORE
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="exercise_frequency",
    y="exam_score"
)

plt.title("Exercise Frequency vs Exam Score")
plt.xlabel("Exercise Frequency")
plt.ylabel("Exam Score")

plt.show()


# ============================================================
# 21. GENDER VS EXAM SCORE
# ============================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="gender",
    y="exam_score"
)

plt.title("Gender vs Exam Score")

plt.show()


# ============================================================
# 22. PART-TIME JOB VS EXAM SCORE
# ============================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="part_time_job",
    y="exam_score"
)

plt.title("Part-Time Job vs Exam Score")

plt.show()


# ============================================================
# 23. DIET QUALITY VS EXAM SCORE
# ============================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="diet_quality",
    y="exam_score"
)

plt.title("Diet Quality vs Exam Score")

plt.show()


# ============================================================
# 24. INTERNET QUALITY VS EXAM SCORE
# ============================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="internet_quality",
    y="exam_score"
)

plt.title("Internet Quality vs Exam Score")

plt.show()


# ============================================================
# 25. OUTLIER DETECTION USING IQR
# ============================================================

outlier_summary = {}

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary[column] = len(outliers)


print("\nNumber of Outliers:")
for column, count in outlier_summary.items():

    print(
        column,
        ":",
        count
    )


# ============================================================
# 26. VISUALIZE OUTLIERS
# ============================================================

for column in numerical_columns:

    plt.figure(figsize=(7, 4))

    sns.boxplot(
        x=df[column]
    )

    plt.title(
        "Boxplot - " + column
    )

    plt.xlabel(column)

    plt.show()


# ============================================================
# 27. PREPARE DATA FOR REGRESSION
# ============================================================

# Remove student_id because it is only an identifier

X = df.drop(
    columns=[
        "exam_score",
        "student_id"
    ]
)

y = df["exam_score"]


print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print("exam_score")


# ============================================================
# 28. IDENTIFY FEATURE TYPES
# ============================================================

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()


print("\nNumerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)


# ============================================================
# 29. TRAIN-TEST SPLIT FOR REGRESSION
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print("\nTraining Data Shape:")
print(X_train.shape)

print("\nTesting Data Shape:")
print(X_test.shape)


# ============================================================
# 30. PREPROCESSING FOR REGRESSION
# ============================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )
    ]
)


categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numerical_features
        ),

        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ============================================================
# 31. LINEAR REGRESSION MODEL
# ============================================================

linear_regression_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),

        (
            "model",
            LinearRegression()
        )
    ]
)


# Train model

linear_regression_model.fit(
    X_train,
    y_train
)


print("\nLinear Regression model trained successfully!")


# ============================================================
# 32. MAKE REGRESSION PREDICTIONS
# ============================================================

y_train_pred = linear_regression_model.predict(
    X_train
)

y_test_pred = linear_regression_model.predict(
    X_test
)


print("\nFirst 10 Test Predictions:")

print(y_test_pred[:10])


# ============================================================
# 33. ACTUAL VS PREDICTED VALUES
# ============================================================

comparison = pd.DataFrame({

    "Actual Score": y_test.values,

    "Predicted Score": y_test_pred

})


print("\nActual vs Predicted:")
print(comparison.head(10))


# ============================================================
# 34. REGRESSION EVALUATION
# ============================================================

# Training metrics

train_mae = mean_absolute_error(
    y_train,
    y_train_pred
)

train_mse = mean_squared_error(
    y_train,
    y_train_pred
)

train_rmse = np.sqrt(
    train_mse
)

train_r2 = r2_score(
    y_train,
    y_train_pred
)


# Testing metrics

test_mae = mean_absolute_error(
    y_test,
    y_test_pred
)

test_mse = mean_squared_error(
    y_test,
    y_test_pred
)

test_rmse = np.sqrt(
    test_mse
)

test_r2 = r2_score(
    y_test,
    y_test_pred
)


print("\n======================================")
print("LINEAR REGRESSION PERFORMANCE")
print("======================================")


print("\nTRAINING PERFORMANCE")
print("--------------------")

print("MAE :", round(train_mae, 3))
print("MSE :", round(train_mse, 3))
print("RMSE:", round(train_rmse, 3))
print("R²  :", round(train_r2, 3))


print("\nTESTING PERFORMANCE")
print("-------------------")

print("MAE :", round(test_mae, 3))
print("MSE :", round(test_mse, 3))
print("RMSE:", round(test_rmse, 3))
print("R²  :", round(test_r2, 3))


# ============================================================
# 35. ACTUAL VS PREDICTED PLOT
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_test_pred
)

plt.xlabel("Actual Exam Score")
plt.ylabel("Predicted Exam Score")

plt.title(
    "Actual vs Predicted Exam Scores"
)

plt.show()


# ============================================================
# 36. REGRESSION LINE
# ============================================================

plt.figure(figsize=(8, 6))

sns.regplot(
    x=y_test,
    y=y_test_pred
)

plt.xlabel("Actual Exam Score")
plt.ylabel("Predicted Exam Score")

plt.title(
    "Actual vs Predicted Regression Plot"
)

plt.show()


# ============================================================
# 37. CREATE PASS/FAIL TARGET
# ============================================================

# Passing mark = 50

df["pass_fail"] = np.where(
    df["exam_score"] >= 50,
    1,
    0
)


print("\nPass/Fail Target:")
print(
    df[
        [
            "exam_score",
            "pass_fail"
        ]
    ].head()
)


# ============================================================
# 38. CHECK PASS/FAIL DISTRIBUTION
# ============================================================

print("\nPass/Fail Counts:")

print(
    df["pass_fail"].value_counts()
)


print("\nPass/Fail Percentages:")

print(
    df["pass_fail"]
    .value_counts(
        normalize=True
    ) * 100
)


# ============================================================
# 39. CLASSIFICATION DATA
# ============================================================

X_classification = df.drop(
    columns=[
        "exam_score",
        "pass_fail",
        "student_id"
    ]
)

y_classification = df["pass_fail"]


print("\nClassification Features:")

print(
    X_classification.columns.tolist()
)


# ============================================================
# 40. IDENTIFY CLASSIFICATION FEATURE TYPES
# ============================================================

numeric_features_c = (
    X_classification
    .select_dtypes(
        include=["int64", "float64"]
    )
    .columns
    .tolist()
)


categorical_features_c = (
    X_classification
    .select_dtypes(
        include=["object"]
    )
    .columns
    .tolist()
)


print("\nNumerical Classification Features:")
print(numeric_features_c)

print("\nCategorical Classification Features:")
print(categorical_features_c)


# ============================================================
# 41. TRAIN-TEST SPLIT FOR CLASSIFICATION
# ============================================================

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_classification,
    y_classification,
    test_size=0.20,
    random_state=42,
    stratify=y_classification
)


print("\nClassification Training Shape:")
print(X_train_c.shape)

print("\nClassification Testing Shape:")
print(X_test_c.shape)


# ============================================================
# 42. CLASSIFICATION PREPROCESSING
# ============================================================

classification_preprocessor = ColumnTransformer(

    transformers=[

        (
            "num",

            Pipeline(
                steps=[
                    (
                        "imputer",
                        SimpleImputer(
                            strategy="median"
                        )
                    ),

                    (
                        "scaler",
                        StandardScaler()
                    )
                ]
            ),

            numeric_features_c
        ),


        (
            "cat",

            Pipeline(
                steps=[
                    (
                        "imputer",
                        SimpleImputer(
                            strategy="most_frequent"
                        )
                    ),

                    (
                        "encoder",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            drop="first"
                        )
                    )
                ]
            ),

            categorical_features_c
        )
    ]
)


# ============================================================
# 43. LOGISTIC REGRESSION MODEL
# ============================================================

classification_model = Pipeline(
    steps=[

        (
            "preprocessor",
            classification_preprocessor
        ),

        (
            "model",
            LogisticRegression(
                max_iter=1000
            )
        )
    ]
)


# Train classification model

classification_model.fit(
    X_train_c,
    y_train_c
)


print("\nLogistic Regression model trained successfully!")


# ============================================================
# 44. CLASSIFICATION PREDICTIONS
# ============================================================

y_train_pred_c = classification_model.predict(
    X_train_c
)

y_test_pred_c = classification_model.predict(
    X_test_c
)


print("\nFirst 20 Test Predictions:")

print(
    y_test_pred_c[:20]
)


# ============================================================
# 45. CLASSIFICATION PERFORMANCE
# ============================================================

# Training

train_accuracy = accuracy_score(
    y_train_c,
    y_train_pred_c
)

train_precision = precision_score(
    y_train_c,
    y_train_pred_c
)

train_recall = recall_score(
    y_train_c,
    y_train_pred_c
)

train_f1 = f1_score(
    y_train_c,
    y_train_pred_c
)


# Testing

test_accuracy = accuracy_score(
    y_test_c,
    y_test_pred_c
)

test_precision = precision_score(
    y_test_c,
    y_test_pred_c
)

test_recall = recall_score(
    y_test_c,
    y_test_pred_c
)

test_f1 = f1_score(
    y_test_c,
    y_test_pred_c
)


print("\n======================================")
print("LOGISTIC REGRESSION PERFORMANCE")
print("======================================")


print("\nTRAINING PERFORMANCE")
print("--------------------")

print(
    "Accuracy :",
    round(train_accuracy, 3)
)

print(
    "Precision:",
    round(train_precision, 3)
)

print(
    "Recall   :",
    round(train_recall, 3)
)

print(
    "F1 Score :",
    round(train_f1, 3)
)


print("\nTESTING PERFORMANCE")
print("-------------------")

print(
    "Accuracy :",
    round(test_accuracy, 3)
)

print(
    "Precision:",
    round(test_precision, 3)
)

print(
    "Recall   :",
    round(test_recall, 3)
)

print(
    "F1 Score :",
    round(test_f1, 3)
)


# ============================================================
# 46. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test_c,
        y_test_pred_c,
        target_names=[
            "Fail",
            "Pass"
        ]
    )
)


# ============================================================
# 47. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test_c,
    y_test_pred_c
)


print("\nConfusion Matrix:")

print(cm)


# ============================================================
# 48. CONFUSION MATRIX VISUALIZATION
# ============================================================

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Fail",
        "Pass"
    ],
    yticklabels=[
        "Fail",
        "Pass"
    ]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.title(
    "Confusion Matrix - Pass/Fail Classification"
)

plt.show()


# ============================================================
# 49. TRAINING VS TESTING REGRESSION PERFORMANCE
# ============================================================

regression_comparison = pd.DataFrame({

    "Metric": [
        "MAE",
        "MSE",
        "RMSE",
        "R² Score"
    ],

    "Training": [
        train_mae,
        train_mse,
        train_rmse,
        train_r2
    ],

    "Testing": [
        test_mae,
        test_mse,
        test_rmse,
        test_r2
    ]
})


print("\n======================================")
print("REGRESSION TRAINING VS TESTING")
print("======================================")

print(
    regression_comparison.round(3)
)


# ============================================================
# 50. TRAINING VS TESTING CLASSIFICATION
# ============================================================

classification_comparison = pd.DataFrame({

    "Metric": [

        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"

    ],

    "Training": [

        train_accuracy,
        train_precision,
        train_recall,
        train_f1

    ],

    "Testing": [

        test_accuracy,
        test_precision,
        test_recall,
        test_f1

    ]
})


print("\n======================================")
print("CLASSIFICATION TRAINING VS TESTING")
print("======================================")

print(
    classification_comparison.round(3)
)


# ============================================================
# 51. CHECK OVERFITTING / UNDERFITTING
# ============================================================

r2_difference = (
    train_r2 -
    test_r2
)

accuracy_difference = (
    train_accuracy -
    test_accuracy
)


print("\n======================================")
print("OVERFITTING / UNDERFITTING ANALYSIS")
print("======================================")


print(
    "\nRegression R² Difference:",
    round(r2_difference, 3)
)

print(
    "Classification Accuracy Difference:",
    round(accuracy_difference, 3)
)


# Regression interpretation

if r2_difference > 0.10:

    print(
        "\nRegression: Possible overfitting."
    )

elif train_r2 < 0.50 and test_r2 < 0.50:

    print(
        "\nRegression: Model may be underfitting."
    )

else:

    print(
        "\nRegression: No strong evidence of overfitting or underfitting."
    )


# Classification interpretation

if accuracy_difference > 0.10:

    print(
        "Classification: Possible overfitting."
    )

elif train_accuracy < 0.70 and test_accuracy < 0.70:

    print(
        "Classification: Model may be underfitting."
    )

else:

    print(
        "Classification: No strong evidence of overfitting or underfitting."
    )


# ============================================================
# 52. AVERAGE EXAM SCORE BY CATEGORICAL VARIABLES
# ============================================================

print("\n======================================")
print("AVERAGE EXAM SCORE BY CATEGORIES")
print("======================================")


print("\nGender:")
print(
    df.groupby("gender")["exam_score"]
    .mean()
    .sort_values(
        ascending=False
    )
)


print("\nDiet Quality:")
print(
    df.groupby("diet_quality")["exam_score"]
    .mean()
    .sort_values(
        ascending=False
    )
)


print("\nInternet Quality:")
print(
    df.groupby("internet_quality")["exam_score"]
    .mean()
    .sort_values(
        ascending=False
    )
)


print("\nPart-Time Job:")
print(
    df.groupby("part_time_job")["exam_score"]
    .mean()
    .sort_values(
        ascending=False
    )
)


print("\nParental Education:")
print(
    df.groupby(
        "parental_education_level"
    )["exam_score"]
    .mean()
    .sort_values(
        ascending=False
    )
)


# ============================================================
# 53. PASS/FAIL PERCENTAGE
# ============================================================

pass_percentage = (
    df["pass_fail"]
    .value_counts(
        normalize=True
    ) * 100
)


print("\n======================================")
print("PASS / FAIL DISTRIBUTION")
print("======================================")


print(
    "Fail Percentage:",
    round(
        pass_percentage.get(0, 0),
        2
    ),
    "%"
)


print(
    "Pass Percentage:",
    round(
        pass_percentage.get(1, 0),
        2
    ),
    "%"
)


# ============================================================
# 54. FINAL FIVE INSIGHTS
# ============================================================

print("\n")
print("=" * 60)
print("FIVE MEANINGFUL INSIGHTS ABOUT STUDENT PERFORMANCE")
print("=" * 60)


print("""
1. STUDY HOURS
   Study hours per day are an important factor associated with
   exam performance. Students who spend more time studying tend
   to achieve higher exam scores.


2. ATTENDANCE
   Attendance percentage shows a relationship with exam scores.
   Regular attendance may help students understand course
   material and improve academic performance.


3. LIFESTYLE HABITS
   Sleep, exercise, social media usage and entertainment habits
   are associated with student performance. Maintaining a
   balanced lifestyle can support better academic outcomes.


4. MENTAL HEALTH
   Mental health rating can be related to academic performance.
   Better mental well-being may support concentration,
   consistency and effective studying.


5. MACHINE LEARNING PREDICTION
   Linear Regression can be used to predict exam scores from
   student academic, lifestyle and demographic characteristics.
   Logistic Regression can classify students into Pass and Fail
   categories. Comparing training and testing metrics helps
   determine how well the models generalize to unseen students.
""")


# ============================================================
# 55. FINAL MODEL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("FINAL MODEL SUMMARY")
print("=" * 60)


print("\nLINEAR REGRESSION")
print("-----------------")

print(
    "Training MAE :",
    round(train_mae, 3)
)

print(
    "Testing MAE  :",
    round(test_mae, 3)
)

print(
    "Training RMSE:",
    round(train_rmse, 3)
)

print(
    "Testing RMSE :",
    round(test_rmse, 3)
)

print(
    "Training R²  :",
    round(train_r2, 3)
)

print(
    "Testing R²   :",
    round(test_r2, 3)
)


print("\nLOGISTIC REGRESSION")
print("-------------------")

print(
    "Training Accuracy :",
    round(train_accuracy, 3)
)

print(
    "Testing Accuracy  :",
    round(test_accuracy, 3)
)

print(
    "Training Precision:",
    round(train_precision, 3)
)

print(
    "Testing Precision :",
    round(test_precision, 3)
)

print(
    "Training Recall   :",
    round(train_recall, 3)
)

print(
    "Testing Recall    :",
    round(test_recall, 3)
)

print(
    "Training F1 Score :",
    round(train_f1, 3)
)

print(
    "Testing F1 Score  :",
    round(test_f1, 3)
)


# ============================================================
# 56. END
# ============================================================

print("\n")
print("=" * 60)
print("MACHINE LEARNING WORKFLOW COMPLETED SUCCESSFULLY")
print("=" * 60)
```
