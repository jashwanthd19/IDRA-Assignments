```python
import random

print("---Number Guessing Number---")

lower=int(input("Enter lower number:"))
upper=int(input("Enter upper number:"))

secret_no=random.randint(lower,upper)

attempts=10

print(f"\nI have chosen a number between {lower} and {upper}.")
print(f"You have {attempts} attempts to guess it.\n")

for i in range(1,attempts+1):
  guess=int(input(f"Attempt:{i}, Enter your guess:"))

  if guess==secret_no:
    print(f"\nCongratulations the correct number is:{secret_no}.")
    print(f"You guessed it in {i} attempts.")
    break
  elif guess<secret_no:
    print("Too low,Try Again!\n")
  else:
    print("Too high,Try Again!\n")

else:
  print(f"\nSorry! You have reached your attempts limit.")
  print(f"The correct number was {secret_no}.")
  print("Good luck next time!")


```
