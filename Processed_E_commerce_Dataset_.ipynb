```python
import pandas as pd

orders = pd.read_csv("Day9_Orders.csv")
customers = pd.read_csv("Day9_Customers.csv")
products = pd.read_csv("Day9_Products.csv")

print("Orders Dataset")
print(orders.head())

print("\nCustomers Dataset")
print(customers.head())

print("\nProducts Dataset")
print(products.head())

print("\nDataset Shapes")
print("Orders:", orders.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)

print("\nMissing Values")
print("Orders:\n", orders.isnull().sum())
print("\nCustomers:\n", customers.isnull().sum())
print("\nProducts:\n", products.isnull().sum())

print("\nDuplicate Rows")
print("Orders:", orders.duplicated().sum())
print("Customers:", customers.duplicated().sum())
print("Products:", products.duplicated().sum())

orders = orders.drop_duplicates()
customers = customers.drop_duplicates()
products = products.drop_duplicates()

for df in [orders, customers, products]:
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()

orders["Order_Date"] = pd.to_datetime(orders["Order_Date"], errors="coerce")
orders["Quantity"] = pd.to_numeric(orders["Quantity"], errors="coerce")
products["Unit_Price"] = pd.to_numeric(products["Unit_Price"], errors="coerce")

print("\nData Types After Refinement")
print(orders.dtypes)
print(products.dtypes)

part1 = orders.iloc[:len(orders)//2]
part2 = orders.iloc[len(orders)//2:]

combined_orders = pd.concat([part1, part2], ignore_index=True)

print("\nDataFrame After concat()")
print(combined_orders.head())
print("Shape:", combined_orders.shape)

merged_data = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left",
    validate="many_to_one"
)

merged_data = pd.merge(
    merged_data,
    products,
    on="Product_ID",
    how="left",
    validate="many_to_one"
)

print("\nMerged Dataset")
print(merged_data.head())

merged_data["Order_Month"] = merged_data["Order_Date"].dt.month
merged_data["Order_Month_Name"] = merged_data["Order_Date"].dt.month_name()
merged_data["Order_Day"] = merged_data["Order_Date"].dt.day
merged_data["Order_Day_Name"] = merged_data["Order_Date"].dt.day_name()

merged_data["Total_Amount"] = merged_data.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

final_data = merged_data[
    [
        "Order_ID",
        "Order_Date",
        "Order_Month",
        "Order_Month_Name",
        "Order_Day",
        "Order_Day_Name",
        "Customer_ID",
        "Customer_Name",
        "City",
        "Region",
        "Membership_Type",
        "Product_ID",
        "Product_Name",
        "Category",
        "Brand",
        "Unit_Price",
        "Quantity",
        "Total_Amount",
        "Payment_Method",
        "Order_Status"
    ]
]

final_data = final_data.drop_duplicates()
final_data = final_data.sort_values("Order_Date")
final_data = final_data.reset_index(drop=True)

print("\nFinal Processed Dataset")
print(final_data.head(10))

print("\nFinal Dataset Shape")
print(final_data.shape)

print("\nFinal Dataset Information")
print(final_data.info())

print("\nFinal Missing Values")
print(final_data.isnull().sum())

print("\nFinal Duplicate Rows")
print(final_data.duplicated().sum())

final_data.to_csv(
    "Processed_Ecommerce_Dataset.csv",
    index=False
)

print("\nProcessed_Ecommerce_Dataset.csv created successfully!")
```
