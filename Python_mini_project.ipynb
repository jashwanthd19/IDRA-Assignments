```python
class Question:
    def __init__(self, question, options, answer):
        self.question = question
        self.options = options
        self.answer = answer

    def display_question(self):
        print("\n" + self.question)

        for i, option in enumerate(self.options, 1):
            print(f"{i}. {option}")

    def check_answer(self, choice):
        return self.options[choice - 1] == self.answer


class Quiz:
    def __init__(self):
        self.questions = [
            Question(
                "What is the capital of India?",
                ("Mumbai", "Delhi", "Chennai", "Kolkata"),
                "Delhi"
            ),
            Question(
                "Which language is used for Python programming?",
                ("Python", "HTML", "CSS", "SQL"),
                "Python"
            ),
            Question(
                "What is 5 + 3?",
                ("6", "7", "8", "9"),
                "8"
            ),
            Question(
                "Which keyword is used to create a class in Python?",
                ("function", "class", "def", "object"),
                "class"
            ),
            Question(
                "Which data type stores True or False?",
                ("int", "float", "bool", "str"),
                "bool"
            )
        ]

        self.score = 0
        self.attempted = set()

    def start_quiz(self, name):
        self.score = 0
        self.attempted.clear()

        print("\n===== QUIZ STARTED =====")
        print("Player:", name)

        for i, question in enumerate(self.questions):
            question.display_question()

            while True:
                try:
                    choice = int(input("Enter your answer (1-4): "))

                    if choice < 1 or choice > 4:
                        print("Please enter a number between 1 and 4.")
                        continue

                    break

                except ValueError:
                    print("Invalid input. Please enter a number.")

            self.attempted.add(i)

            if question.check_answer(choice):
                print("Correct!")
                self.score += 1
            else:
                print("Wrong!")
                print("Correct answer:", question.answer)

        self.display_result(name)

    def display_result(self, name):
        total = len(self.questions)
        percentage = (self.score / total) * 100

        if percentage >= 50:
            result = "PASS"
        else:
            result = "FAIL"

        print("\n===== QUIZ RESULT =====")
        print("Name:", name)
        print("Score:", self.score, "/", total)
        print("Percentage:", percentage, "%")
        print("Result:", result)

        self.save_result(name, percentage, result)

    def save_result(self, name, percentage, result):
        try:
            with open("quiz_results.txt", "a") as file:
                file.write(
                    f"Name: {name}, "
                    f"Score: {self.score}/{len(self.questions)}, "
                    f"Percentage: {percentage}%, "
                    f"Result: {result}\n"
                )

            print("Result saved successfully.")

        except Exception:
            print("Error while saving result.")

    def view_results(self):
        try:
            with open("quiz_results.txt", "r") as file:
                data = file.read()

                if data:
                    print("\n===== PREVIOUS RESULTS =====")
                    print(data)
                else:
                    print("\nNo previous results found.")

        except FileNotFoundError:
            print("\nNo previous results found.")


def main():
    quiz = Quiz()

    while True:
        print("\n==========================")
        print("     QUIZ APPLICATION")
        print("==========================")
        print("1. Start Quiz")
        print("2. View Previous Results")
        print("3. Exit")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:
                name = input("Enter your name: ")

                if name.strip() == "":
                    print("Name cannot be empty.")
                else:
                    quiz.start_quiz(name)

            elif choice == 2:
                quiz.view_results()

            elif choice == 3:
                print("Thank you for using the Quiz Application!")
                break

            else:
                print("Invalid choice. Please select 1, 2, or 3.")

        except ValueError:
            print("Invalid input. Please enter a number.")


main()
```
