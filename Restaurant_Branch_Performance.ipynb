```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("Day13_Restaurant_Branch_Performance_Dataset.csv")


print("Dataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Information:")
df.info()


print("\nSummary Statistics:")
print(df.describe())


print("\nCategorical Summary:")
print(df.describe(include="object"))


print("\nCustomers:")
print(df["Customers"].describe())

print("\nRevenue:")
print(df["Revenue"].describe())

print("\nProfit:")
print(df["Profit"].describe())

print("\nMarketing Spend:")
print(df["Marketing_Spend"].describe())

print("\nStaff Count:")
print(df["Staff_Count"].describe())

print("\nDelivery Time:")
print(df["Avg_Delivery_Min"].describe())

print("\nCustomer Rating:")
print(df["Customer_Rating"].describe())


print("\nBranch Performance:")
branch_performance = df.groupby("Branch").agg({
    "Customers": "mean",
    "Revenue": "mean",
    "Profit": "mean",
    "Marketing_Spend": "mean",
    "Staff_Count": "mean",
    "Avg_Delivery_Min": "mean",
    "Customer_Rating": "mean"
}).round(2)

print(branch_performance)


print("\nRegion Performance:")
region_performance = df.groupby("Region").agg({
    "Customers": "mean",
    "Revenue": "mean",
    "Profit": "mean",
    "Marketing_Spend": "mean",
    "Staff_Count": "mean",
    "Avg_Delivery_Min": "mean",
    "Customer_Rating": "mean"
}).round(2)

print(region_performance)


print("\nStore Type Performance:")
store_performance = df.groupby("Store_Type").agg({
    "Customers": "mean",
    "Revenue": "mean",
    "Profit": "mean",
    "Marketing_Spend": "mean",
    "Staff_Count": "mean",
    "Avg_Delivery_Min": "mean",
    "Customer_Rating": "mean"
}).round(2)

print(store_performance)


plt.figure(figsize=(8, 5))
sns.histplot(df["Customers"], bins=20, kde=True)
plt.title("Distribution of Customers")
plt.xlabel("Customers")
plt.ylabel("Frequency")
plt.show()


plt.figure(figsize=(8, 5))
sns.histplot(df["Revenue"], bins=20, kde=True)
plt.title("Distribution of Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()


plt.figure(figsize=(8, 5))
sns.histplot(df["Profit"], bins=20, kde=True)
plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.show()


plt.figure(figsize=(8, 5))
sns.histplot(df["Marketing_Spend"], bins=20, kde=True)
plt.title("Distribution of Marketing Spend")
plt.xlabel("Marketing Spend")
plt.ylabel("Frequency")
plt.show()


plt.figure(figsize=(8, 5))
sns.histplot(df["Staff_Count"], bins=15, kde=True)
plt.title("Distribution of Staff Count")
plt.xlabel("Staff Count")
plt.ylabel("Frequency")
plt.show()


plt.figure(figsize=(8, 5))
sns.histplot(df["Avg_Delivery_Min"], bins=20, kde=True)
plt.title("Distribution of Delivery Time")
plt.xlabel("Average Delivery Time")
plt.ylabel("Frequency")
plt.show()


plt.figure(figsize=(8, 5))
sns.histplot(df["Customer_Rating"], bins=15, kde=True)
plt.title("Distribution of Customer Rating")
plt.xlabel("Customer Rating")
plt.ylabel("Frequency")
plt.show()


branch_revenue = df.groupby("Branch")["Revenue"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
branch_revenue.plot(kind="bar")
plt.title("Average Revenue by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Revenue")
plt.xticks(rotation=45)
plt.show()


branch_profit = df.groupby("Branch")["Profit"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
branch_profit.plot(kind="bar")
plt.title("Average Profit by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Profit")
plt.xticks(rotation=45)
plt.show()


region_revenue = df.groupby("Region")["Revenue"].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
region_revenue.plot(kind="bar")
plt.title("Average Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Average Revenue")
plt.show()


region_profit = df.groupby("Region")["Profit"].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
region_profit.plot(kind="bar")
plt.title("Average Profit by Region")
plt.xlabel("Region")
plt.ylabel("Average Profit")
plt.show()


store_revenue = df.groupby("Store_Type")["Revenue"].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
store_revenue.plot(kind="bar")
plt.title("Average Revenue by Store Type")
plt.xlabel("Store Type")
plt.ylabel("Average Revenue")
plt.show()


store_profit = df.groupby("Store_Type")["Profit"].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
store_profit.plot(kind="bar")
plt.title("Average Profit by Store Type")
plt.xlabel("Store Type")
plt.ylabel("Average Profit")
plt.show()


numeric_columns = [
    "Staff_Count",
    "Marketing_Spend",
    "Customers",
    "Orders",
    "Average_Bill",
    "Revenue",
    "Food_Cost",
    "Operating_Cost",
    "Profit",
    "Avg_Delivery_Min",
    "Customer_Rating",
    "Temperature_C",
    "Humidity_Percent"
]

correlation_matrix = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
print(correlation_matrix.round(2))


plt.figure(figsize=(12, 9))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Variables")
plt.show()


print("\nCorrelation with Revenue:")
print(
    correlation_matrix["Revenue"]
    .sort_values(ascending=False)
    .round(2)
)


print("\nCorrelation with Profit:")
print(
    correlation_matrix["Profit"]
    .sort_values(ascending=False)
    .round(2)
)


print("\nTop Branch by Average Revenue:")
print(branch_revenue.index[0], branch_revenue.iloc[0])


print("\nTop Branch by Average Profit:")
print(branch_profit.index[0], branch_profit.iloc[0])


print("\nTop Region by Average Revenue:")
print(region_revenue.index[0], region_revenue.iloc[0])


print("\nTop Region by Average Profit:")
print(region_profit.index[0], region_profit.iloc[0])


print("\nTop Store Type by Average Revenue:")
print(store_revenue.index[0], store_revenue.iloc[0])


print("\nTop Store Type by Average Profit:")
print(store_profit.index[0], store_profit.iloc[0])
```
