```python
# ============================================================
# STUDENT WELLBEING STATISTICAL ANALYSIS & PROBABILITY
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES AND LOAD DATASET
# ============================================================

import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("Day16_Student_Wellbeing_Survey.csv")

print("=" * 60)
print("DATASET INSPECTION")
print("=" * 60)

print("\nShape of dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nRandom 5 rows:")
display(df.sample(5))

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())


# ============================================================
# 2. DESCRIPTIVE STATISTICS
# Mean, Median and Mode
# ============================================================

variables = [
    "Weekly_Study_Hours",
    "Average_Sleep_Hours",
    "Daily_Screen_Time_Hours",
    "Stress_Score",
    "Academic_Readiness_Score"
]

print("\n")
print("=" * 60)
print("MEAN, MEDIAN AND MODE")
print("=" * 60)

for col in variables:

    print("\nVariable:", col)

    print("Mean   :", round(df[col].mean(), 3))
    print("Median :", round(df[col].median(), 3))
    print("Mode   :", df[col].mode().tolist())


# ============================================================
# 3. RANGE, VARIANCE, STANDARD DEVIATION, Q1, Q3 AND IQR
# ============================================================

print("\n")
print("=" * 60)
print("RANGE, VARIANCE, STANDARD DEVIATION, Q1, Q3 AND IQR")
print("=" * 60)

statistics = []

for col in variables:

    minimum = df[col].min()
    maximum = df[col].max()

    range_value = maximum - minimum

    variance = df[col].var()

    standard_deviation = df[col].std()

    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)

    iqr = q3 - q1

    statistics.append({
        "Variable": col,
        "Mean": df[col].mean(),
        "Median": df[col].median(),
        "Mode": df[col].mode().tolist(),
        "Range": range_value,
        "Variance": variance,
        "Standard Deviation": standard_deviation,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr
    })

stats_df = pd.DataFrame(statistics)

display(stats_df.round(3))


# ============================================================
# 4. VARIABLE WITH GREATEST VARIABILITY
# ============================================================

print("\n")
print("=" * 60)
print("GREATEST VARIABILITY")
print("=" * 60)

max_std_index = stats_df["Standard Deviation"].idxmax()

max_variability_variable = stats_df.loc[
    max_std_index, "Variable"
]

max_std = stats_df.loc[
    max_std_index, "Standard Deviation"
]

print("Variable with greatest variability:",
      max_variability_variable)

print("Standard Deviation:",
      round(max_std, 3))

print("\nInterpretation:")
print(
    max_variability_variable,
    "has the greatest absolute variability based on standard deviation."
)


# ============================================================
# 5. IQR OUTLIER DETECTION
# ============================================================

outlier_variables = [
    "Weekly_Study_Hours",
    "Daily_Screen_Time_Hours",
    "Commute_Time_Minutes",
    "Monthly_Discretionary_Spending"
]

print("\n")
print("=" * 60)
print("IQR OUTLIER DETECTION")
print("=" * 60)

outlier_summary = []

for col in outlier_variables:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_fence = Q1 - 1.5 * IQR
    upper_fence = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_fence) |
        (df[col] > upper_fence)
    ]

    outlier_summary.append({
        "Variable": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Fence": lower_fence,
        "Upper Fence": upper_fence,
        "Number of Outliers": len(outliers)
    })

    print("\nVariable:", col)

    print("Q1:", round(Q1, 3))
    print("Q3:", round(Q3, 3))
    print("IQR:", round(IQR, 3))

    print("Lower Fence:",
          round(lower_fence, 3))

    print("Upper Fence:",
          round(upper_fence, 3))

    print("Number of Outliers:",
          len(outliers))

    print("Outlier Values:")
    print(outliers[col].tolist())


outlier_df = pd.DataFrame(outlier_summary)

print("\nOutlier Summary:")
display(outlier_df.round(3))


# ============================================================
# 6. BEFORE AND AFTER REMOVING OUTLIERS
# Using Weekly_Study_Hours
# ============================================================

print("\n")
print("=" * 60)
print("BEFORE AND AFTER OUTLIER REMOVAL")
print("=" * 60)

col = "Weekly_Study_Hours"

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)

IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

# Original data
original_mean = df[col].mean()
original_median = df[col].median()

# Remove outliers
clean_data = df[
    (df[col] >= lower_fence) &
    (df[col] <= upper_fence)
]

clean_mean = clean_data[col].mean()
clean_median = clean_data[col].median()

print("Variable:", col)

print("\nBefore removing outliers:")
print("Mean:", round(original_mean, 3))
print("Median:", round(original_median, 3))

print("\nAfter removing outliers:")
print("Mean:", round(clean_mean, 3))
print("Median:", round(clean_median, 3))

print("\nChange in Mean:",
      round(clean_mean - original_mean, 3))

print("Change in Median:",
      round(clean_median - original_median, 3))


# ============================================================
# 7. DEFINE PROBABILITY EVENTS
# ============================================================

print("\n")
print("=" * 60)
print("PROBABILITY EVENTS")
print("=" * 60)

# Event A
# Student has Part_Time_Job = Yes
A = df["Part_Time_Job"] == "Yes"

# Event B
# Student has Stress_Score >= 7
B = df["Stress_Score"] >= 7

# Event C
# Student has Scholarship = Yes
C = df["Scholarship"] == "Yes"

# Event D
# Student exercises at least 3 days per week
D = df["Exercise_Days_Per_Week"] >= 3

print("Event A: Part-Time Job = Yes")
print("Event B: Stress Score >= 7")
print("Event C: Scholarship = Yes")
print("Event D: Exercise Days >= 3")


# ============================================================
# 8. BASIC PROBABILITIES
# P(A), P(B), P(C), P(D)
# ============================================================

print("\n")
print("=" * 60)
print("BASIC PROBABILITIES")
print("=" * 60)

P_A = A.mean()
P_B = B.mean()
P_C = C.mean()
P_D = D.mean()

print("P(A) =", round(P_A, 4),
      "=", round(P_A * 100, 2), "%")

print("P(B) =", round(P_B, 4),
      "=", round(P_B * 100, 2), "%")

print("P(C) =", round(P_C, 4),
      "=", round(P_C * 100, 2), "%")

print("P(D) =", round(P_D, 4),
      "=", round(P_D * 100, 2), "%")


# ============================================================
# 9. P(A OR B)
# ============================================================

print("\n")
print("=" * 60)
print("P(A OR B)")
print("=" * 60)

# Formula:
# P(A or B) = P(A) + P(B) - P(A and B)

P_A_or_B = (A | B).mean()

print("P(A or B) =",
      round(P_A_or_B, 4))

print("Percentage =",
      round(P_A_or_B * 100, 2), "%")


# ============================================================
# 10. P(A AND B)
# ============================================================

print("\n")
print("=" * 60)
print("P(A AND B)")
print("=" * 60)

P_A_and_B = (A & B).mean()

print("P(A and B) =",
      round(P_A_and_B, 4))

print("Percentage =",
      round(P_A_and_B * 100, 2), "%")


# ============================================================
# 11. P(A | B)
# ============================================================

print("\n")
print("=" * 60)
print("P(A | B)")
print("=" * 60)

# Formula:
# P(A|B) = P(A and B) / P(B)

P_A_given_B = (A & B).sum() / B.sum()

print("P(A | B) =",
      round(P_A_given_B, 4))

print("Percentage =",
      round(P_A_given_B * 100, 2), "%")


# ============================================================
# 12. P(B | A)
# ============================================================

print("\n")
print("=" * 60)
print("P(B | A)")
print("=" * 60)

# Formula:
# P(B|A) = P(A and B) / P(A)

P_B_given_A = (A & B).sum() / A.sum()

print("P(B | A) =",
      round(P_B_given_A, 4))

print("Percentage =",
      round(P_B_given_A * 100, 2), "%")


# ============================================================
# 13. MUTUALLY EXCLUSIVE EVENTS
# Year 1 and Year 4
# ============================================================

print("\n")
print("=" * 60)
print("MUTUALLY EXCLUSIVE EVENTS")
print("=" * 60)

Year1 = df["Year_of_Study"] == 1
Year4 = df["Year_of_Study"] == 4

intersection = Year1 & Year4

print("Number of Year 1 students:",
      Year1.sum())

print("Number of Year 4 students:",
      Year4.sum())

print("Students in both Year 1 and Year 4:",
      intersection.sum())

P_year1_and_year4 = intersection.mean()

print("P(Year 1 AND Year 4) =",
      P_year1_and_year4)

if intersection.sum() == 0:
    print("\nConclusion:")
    print("Year 1 and Year 4 are mutually exclusive.")
else:
    print("\nConclusion:")
    print("Year 1 and Year 4 are not mutually exclusive.")


# ============================================================
# 14. CHECK INDEPENDENCE OF A AND B
# ============================================================

print("\n")
print("=" * 60)
print("INDEPENDENCE OF A AND B")
print("=" * 60)

# For independent events:
# P(A and B) = P(A) * P(B)

P_A_and_B = (A & B).mean()

P_A_times_P_B = P_A * P_B

print("P(A and B) =",
      round(P_A_and_B, 4))

print("P(A) × P(B) =",
      round(P_A_times_P_B, 4))

difference = P_A_and_B - P_A_times_P_B

print("Difference =",
      round(difference, 4))

if np.isclose(P_A_and_B, P_A_times_P_B):
    print("\nConclusion:")
    print("A and B appear to be independent.")
else:
    print("\nConclusion:")
    print("A and B do not appear to be independent.")

    print(
        "\nSince P(A and B) is different from "
        "P(A) × P(B), the events are not independent "
        "in this dataset."
    )


# ============================================================
# 15. BAYES' THEOREM
# ============================================================

print("\n")
print("=" * 60)
print("BAYES' THEOREM")
print("=" * 60)

# We need:
# P(A)
# P(B|A)
# P(B|not A)

not_A = ~A

P_not_A = not_A.mean()

P_B_given_not_A = (
    (B & not_A).sum()
    / not_A.sum()
)

print("P(A) =",
      round(P_A, 4))

print("P(B | A) =",
      round(P_B_given_A, 4))

print("P(B | not A) =",
      round(P_B_given_not_A, 4))

print("P(not A) =",
      round(P_not_A, 4))


# Bayes Formula:
#
# P(A|B) =
# [P(B|A) P(A)]
# -------------------------------
# [P(B|A) P(A)] +
# [P(B|not A) P(not A)]

numerator = P_B_given_A * P_A

denominator = (
    P_B_given_A * P_A
    +
    P_B_given_not_A * P_not_A
)

P_A_given_B_bayes = numerator / denominator

print("\nUsing Bayes' theorem:")

print("Numerator =",
      round(numerator, 6))

print("Denominator =",
      round(denominator, 6))

print("P(A | B) using Bayes =",
      round(P_A_given_B_bayes, 4))


# ============================================================
# 16. VERIFY BAYES WITH DIRECT CALCULATION
# ============================================================

print("\n")
print("=" * 60)
print("BAYES VERIFICATION")
print("=" * 60)

print("Direct P(A | B) =",
      round(P_A_given_B, 4))

print("Bayes P(A | B) =",
      round(P_A_given_B_bayes, 4))

if np.isclose(P_A_given_B, P_A_given_B_bayes):
    print("\nBoth values agree.")
    print("Bayes' theorem is verified.")
else:
    print("\nValues do not agree.")


# ============================================================
# 17. ACADEMIC READINESS SCORE
# MEAN AND STANDARD DEVIATION
# ============================================================

print("\n")
print("=" * 60)
print("ACADEMIC READINESS SCORE")
print("=" * 60)

academic = df["Academic_Readiness_Score"]

mean_score = academic.mean()

std_score = academic.std()

print("Mean =", round(mean_score, 3))

print("Standard Deviation =",
      round(std_score, 3))


# ============================================================
# 18. HIGHEST SCORE AND Z-SCORE
# ============================================================

print("\n")
print("=" * 60)
print("HIGHEST ACADEMIC READINESS SCORE")
print("=" * 60)

highest_score = academic.max()

highest_index = academic.idxmax()

highest_student = df.loc[
    highest_index, "Student_ID"
]

highest_z = (
    highest_score - mean_score
) / std_score

print("Student ID:",
      highest_student)

print("Highest Score:",
      highest_score)

print("Z-score:",
      round(highest_z, 3))

print("\nInterpretation:")

if highest_z > 0:
    print(
        "The highest-scoring student is above the "
        "mean by approximately",
        round(highest_z, 2),
        "standard deviations."
    )
else:
    print(
        "The highest-scoring student is below the mean."
    )


# ============================================================
# 19. LOWEST SCORE AND Z-SCORE
# ============================================================

print("\n")
print("=" * 60)
print("LOWEST ACADEMIC READINESS SCORE")
print("=" * 60)

lowest_score = academic.min()

lowest_index = academic.idxmin()

lowest_student = df.loc[
    lowest_index, "Student_ID"
]

lowest_z = (
    lowest_score - mean_score
) / std_score

print("Student ID:",
      lowest_student)

print("Lowest Score:",
      lowest_score)

print("Z-score:",
      round(lowest_z, 3))

print("\nInterpretation:")

if lowest_z < 0:
    print(
        "The lowest-scoring student is below the "
        "mean by approximately",
        round(abs(lowest_z), 2),
        "standard deviations."
    )
else:
    print(
        "The lowest-scoring student is above the mean."
    )


# ============================================================
# 20. 68-95-99.7 EMPIRICAL RULE
# ============================================================

print("\n")
print("=" * 60)
print("68-95-99.7 EMPIRICAL RULE")
print("=" * 60)

# Within 1 standard deviation
lower_1 = mean_score - std_score
upper_1 = mean_score + std_score

# Within 2 standard deviations
lower_2 = mean_score - 2 * std_score
upper_2 = mean_score + 2 * std_score

# Within 3 standard deviations
lower_3 = mean_score - 3 * std_score
upper_3 = mean_score + 3 * std_score

print("\nWithin 1 Standard Deviation:")
print(
    round(lower_1, 3),
    "to",
    round(upper_1, 3)
)
print("Expected percentage: 68%")

print("\nWithin 2 Standard Deviations:")
print(
    round(lower_2, 3),
    "to",
    round(upper_2, 3)
)
print("Expected percentage: 95%")

print("\nWithin 3 Standard Deviations:")
print(
    round(lower_3, 3),
    "to",
    round(upper_3, 3)
)
print("Expected percentage: 99.7%")


# ============================================================
# 21. ACTUAL PERCENTAGE OF STUDENTS WITHIN 1, 2 AND 3 SD
# ============================================================

print("\n")
print("=" * 60)
print("ACTUAL SAMPLE PERCENTAGES")
print("=" * 60)

within_1_sd = academic[
    (academic >= lower_1) &
    (academic <= upper_1)
]

within_2_sd = academic[
    (academic >= lower_2) &
    (academic <= upper_2)
]

within_3_sd = academic[
    (academic >= lower_3) &
    (academic <= upper_3)
]

percentage_1 = len(within_1_sd) / len(academic) * 100

percentage_2 = len(within_2_sd) / len(academic) * 100

percentage_3 = len(within_3_sd) / len(academic) * 100

print(
    "Actual percentage within 1 SD:",
    round(percentage_1, 2), "%"
)

print(
    "Actual percentage within 2 SD:",
    round(percentage_2, 2), "%"
)

print(
    "Actual percentage within 3 SD:",
    round(percentage_3, 2), "%"
)


# ============================================================
# 22. FINAL PROBABILITY SUMMARY TABLE
# ============================================================

print("\n")
print("=" * 60)
print("FINAL PROBABILITY SUMMARY")
print("=" * 60)

probability_summary = pd.DataFrame({
    "Event": [
        "P(A)",
        "P(B)",
        "P(C)",
        "P(D)",
        "P(A or B)",
        "P(A and B)",
        "P(A | B)",
        "P(B | A)"
    ],

    "Probability": [
        P_A,
        P_B,
        P_C,
        P_D,
        P_A_or_B,
        P_A_and_B,
        P_A_given_B,
        P_B_given_A
    ]
})

probability_summary["Percentage"] = (
    probability_summary["Probability"] * 100
)

display(
    probability_summary.round(4)
)


# ============================================================
# 23. FINAL STATISTICAL OBSERVATIONS
# ============================================================

print("\n")
print("=" * 60)
print("5 MEANINGFUL STATISTICAL OBSERVATIONS")
print("=" * 60)

print("\n1. Academic Readiness Score:")
print(
    "Academic Readiness Score has the greatest absolute "
    "variability among the requested variables based on "
    "standard deviation."
)

print("\n2. Outliers:")
print(
    "IQR analysis shows that extreme observations are "
    "present in study hours, screen time, commute time, "
    "and discretionary spending."
)

print("\n3. Part-Time Job and Stress:")
print(
    "P(A and B) is different from P(A) × P(B), indicating "
    "that having a part-time job and having high stress "
    "do not appear to be independent in this sample."
)

print("\n4. Exercise:")
print(
    "Event D represents students who exercise at least "
    "3 days per week. Its probability shows how common "
    "regular exercise is among the surveyed students."
)

print("\n5. Academic Readiness Distribution:")
print(
    "The Z-scores identify how far the highest and lowest "
    "academic readiness scores are from the sample mean. "
    "The empirical rule provides expected percentages "
    "within 1, 2 and 3 standard deviations."
)


# ============================================================
# 24. FINAL CONCLUSION
# ============================================================

print("\n")
print("=" * 60)
print("FINAL CONCLUSION")
print("=" * 60)

print("""
The Student Wellbeing Survey was analyzed using descriptive
statistics, outlier detection, probability, conditional
probability, independence, Bayes' theorem and the normal
distribution.

The descriptive analysis summarizes the central tendency
and variability of study hours, sleep hours, screen time,
stress score and academic readiness.

The IQR method identifies extreme observations in the
requested variables. Comparing the mean and median before
and after removing outliers demonstrates the effect of
extreme values on statistical measures.

Probability analysis was used to study part-time employment,
high stress, scholarships and exercise. Year 1 and Year 4
were found to be mutually exclusive events.

The independence test compares P(A and B) with
P(A) × P(B). Bayes' theorem was then used to calculate
P(A|B) and verify the direct conditional probability.

Finally, Academic Readiness Score was analyzed using
Z-scores and the 68-95-99.7 empirical rule.
""")
```
